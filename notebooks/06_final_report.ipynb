## 14. Limitations

- Records are listings, not confirmed sale transactions.
- Vehicle condition, location, listing date, trim, and optional equipment are unavailable.
- There is no unique listing identifier.
- Brands have unequal sample sizes.
- Cross-brand correlation can be confounded by brand and model mix.
- Pearson correlation is sensitive to outliers.
- Vehicle age is fixed as of 2024.
- Currency and unit assumptions come from the dataset context.
- Zero tax is not assigned a policy explanation without separate evidence.

## 15. Recommendations

1. Report listing counts with category-level price comparisons.
2. Use medians alongside means for skewed price distributions.
3. Treat mileage-price correlation as descriptive.
4. Interpret low-count age and transmission groups cautiously.
5. Preserve the raw hash manifest for reproducibility.
6. Obtain transaction, condition, location, and listing-date data before making market claims.
