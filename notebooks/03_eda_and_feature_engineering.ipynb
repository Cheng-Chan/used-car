This notebook consumes the Phase 3 canonical dataset. It does not reimplement source cleaning or integration.

## Methodology

Vehicle age is calculated exactly as `2024 - year` using vectorized Pandas arithmetic. Numeric summaries include count, mean, median, standard deviation, minimum, quartiles, maximum, IQR, and counts outside the 1.5-IQR bounds. Outliers are described, not removed.

## Interpretation

The price mean is higher than the median, indicating right-skew in listing prices. High-price listings and unequal brand/model mixes can pull the mean upward, so the median is often a useful description of a typical listing. IQR flags identify values for review but do not establish that they are errors.

Grouped summaries describe this dataset's listings. They do not establish that brand, mileage, vehicle age, or any other single variable causes price differences.

## Data-quality or analytical limitations

The vehicle-age feature is fixed as of 2024 and is not the current age. Listings are not confirmed sales, and condition, location, listing date, trim, and optional equipment are unavailable. Brand sample sizes are unequal, and the observed outlier counts reflect the source data rather than removed records.

## Findings and Decisions

- Keep the fixed formula `vehicle_age = 2024 - year`.
- Preserve integer vehicle-age values and reject negative ages.
- Keep outliers in the enriched dataset and report their IQR counts.
- Use medians alongside means when distributions are skewed.
- Carry these descriptive tables into the guiding-exercise phase.
