# Matplotlib Visualizations

## Objective

Create separate, presentation-ready charts for the required analytical results while preserving the complete-data calculations and documenting visual limitations.

## Questions being answered

- How do average prices compare for Petrol, Diesel, and Hybrid vehicles?
- What does the full-data mileage-price relationship look like?
- How do transmission counts and median prices differ?
- How are prices distributed across brands?
- Which models are most common, and how does average price vary by vehicle age?
- How imbalanced are the brand listing counts?

## Imports

```python
import pandas as pd
from IPython.display import Image, display

from src.analysis import build_guiding_exercise_outputs
from src.config import FIGURES_DIR, PROCESSED_DATA_DIR
from src.plotting import create_required_figures
```

## Data-loading section

The notebook consumes the Phase 4 enriched dataset and Phase 5 reusable analysis functions. All charts are produced through reusable functions in `src.plotting`.

```python
dataframe = pd.read_csv(PROCESSED_DATA_DIR / "used_cars_enriched.csv")
outputs = build_guiding_exercise_outputs(dataframe)
figure_paths = create_required_figures(dataframe, outputs)
print(f"Created {len(figure_paths)} figures in {FIGURES_DIR}.")
```

## Methodology

Matplotlib is used directly. Figures are saved separately at 220 DPI with tight bounding boxes. The mileage-price chart plots all valid records and uses rasterization for readability. The brand box plot hides fliers only in the display; no observations are removed from the dataset or calculations.

## Average price by fuel type

```python
display(Image(filename=str(figure_paths["average_price_by_fuel_type.png"])))
```

The chart compares average listing prices for the three required fuel types and annotates their listing counts. It does not imply that the category with the highest average is the most common or that fuel type alone causes price differences.

## Mileage versus price

```python
display(Image(filename=str(figure_paths["mileage_vs_price_scatter.png"])))
```

The full-data scatter plot shows a broad negative association between mileage and price. The plotted relationship is descriptive and does not establish causation or function as a predictive model.

## Transmission charts

```python
display(Image(filename=str(figure_paths["car_count_by_transmission.png"])))
display(Image(filename=str(figure_paths["median_price_by_transmission.png"])))
```

The two charts use separate units: one shows listing counts and the other shows median GBP price. They should not be read as the same measure or as evidence that transmission causes the observed prices.

## Price distribution across brands

```python
display(Image(filename=str(figure_paths["price_distribution_by_brand.png"])))
```

The box plot compares the central ranges of brand prices ordered by median. Fliers are hidden only to keep the chart readable; the underlying rows remain available and were retained in all calculations.

## Most common models

```python
display(Image(filename=str(figure_paths["top_three_models.png"])))
```

The chart shows the three most common raw model labels. Model names can exist across multiple brands, so the accompanying Phase 5 brand-model table should be used when brand identity matters.

## Average price by vehicle age

```python
display(Image(filename=str(figure_paths["average_price_by_vehicle_age.png"])))
```

The chart shows average listing price by age as of the fixed 2024 reference year. Age groups with few listings may be unstable, so counts in `price_by_vehicle_age.csv` should be consulted alongside the line.

## Listings by brand

```python
display(Image(filename=str(figure_paths["listing_count_by_brand.png"])))
```

The brand count chart makes the dataset imbalance visible. Cross-brand comparisons should account for unequal sample sizes and differences in model mix.

## Data-quality or analytical limitations

Charts summarize listing data rather than confirmed transactions. Price may reflect condition, location, trim, optional equipment, and listing date, which are not available. Visual choices such as hidden box-plot fliers improve readability but do not remove data.

## Findings and Decisions

- Save each major chart separately.
- Use GBP labels rather than dollar labels.
- Plot all valid mileage-price records.
- Keep count and median transmission charts separate.
- State when visual outlier points are hidden.
- Preserve full analytical tables for the final report.
